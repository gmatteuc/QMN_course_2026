## 5. Variables, basic types and functions

A **variable** is a *name* you give to a *value*, so you can reuse the value later in your code. Think of it as a labelled box: the **label** is the name (on the left of `=`), and the **box** holds the value (on the right of `=`).

Let's create the simplest variable possible: a counter that holds the number `240`.


````python
# --- Our very first variable
n_trials = 240    # the value 240 is assigned to the name "n_trials"
print(n_trials)   # later we can ask Python to print whatever n_trials currently holds
````

So far we used an integer (`240`). But a variable can hold many other kinds of values: decimal numbers, text, truth values, or "nothing" (a placeholder). The kind of value is called the **type**. You can ask Python what type a value has with the built-in `type()` function. Let's create a variable of each common type and inspect it:


````python
# --- Some variables of different types
n_subjects = 12             # int   : how many animals took part
mean_rt    = 0.485          # float : a reaction time, in seconds
subject_id = "S01"          # str   : a label for one animal, made up here
is_correct = True           # bool  : did it get the trial right
not_set    = None           # NoneType: nothing recorded yet

print(n_subjects, type(n_subjects))
print(mean_rt,    type(mean_rt))
print(subject_id, type(subject_id))
print(is_correct, type(is_correct))
print(not_set,    type(not_set))
````

**Reference: the basic types you'll meet in this course:**

| Type        | Example             | Represents                |
|-------------|---------------------|---------------------------|
| `int`       | `12`                | integer numbers           |
| `float`     | `3.14`              | real numbers (decimal)    |
| `str`       | `"hello"`           | text (a "string")         |
| `bool`      | `True` / `False`    | truth values (booleans)   |
| `NoneType`  | `None`              | "nothing here yet"        |

**Naming conventions.** Use **descriptive** names in `snake_case` (lowercase words separated by underscores): `mean_rt`, `n_trials`, `subject_id`. The name of a variable should ideally tell something about its meaning (this will make your code more readable by you and others). Avoid single letters except for short throwaway counters like `i`, `j`. It is also good practice to comment the code you write describing in words what each line does (after a `#` symbol in Python).


You can **reassign** a variable at any time: Python does not care what type the new value has:

````python
# --- A variable can be reassigned (and its type can even change)
x = 10 # after this line, x holds the integer value 10
print(x, type(x)) # print the value and type of x
x = "ten" # after this line, x holds the string value "ten"
print(x, type(x)) # print the value and type of x again (notice how it changed)
````

### 6.2 Strings and f-strings

A **string** is a sequence of characters surrounded by quotes (single `'…'` or double `"…"`, both work; pick one and stay consistent).

The single most useful tool for *combining text with values* is the **f-string**: a string prefixed with `f` in which any `{expression}` inside braces is replaced by the value of that expression. You will use f-strings constantly:


````python
# --- f-strings
subject = "S01"      # one animal identifier
n_trials = 1440      # how many trials it did
mean_rt  = 0.4853    # its average reaction time, in seconds

# Anything inside {...} inside an f-string is replaced by its value:
print(f"Subject {subject} ran {n_trials} trials with mean RT {mean_rt:.3f} s.")

# The :.3f after mean_rt means "format as a float with 3 decimals".
# Other useful format specifiers:
print(f"{mean_rt:.0f}")     # round to nearest integer
print(f"{mean_rt * 100:.1f}%")  # percentage, 1 decimal
print(f"{n_trials:,}")      # thousands separator
print(f"{0.826:.1%}")       # a fraction written as a percentage, one decimal
````

## 7. Collections: lists, tuples, dictionaries

Real data are rarely a single number: you usually have *many* of them, often labelled. Python ships with three workhorse containers.

### 7.1 Lists

A **list** is an ordered, mutable (= modifiable) collection of values. You write it with **square brackets** and commas. Lists can contain anything, even values of different types. Write the name of the list followed by a position in square brackets to read the value stored there. Remember that Python follows a 0-indexing convention, **counting from 0 rather than 1**, so the first element sits at position 0.


````python
# --- Lists
subjects = ["S01", "S02", "S03"]          # three animals, with made-up labels
rts      = [0.41, 0.55, 0.62, 0.39, 0.48]  # five reaction times, in seconds

print(subjects)
print("length:", len(subjects))          # number of elements
print("first :", subjects[0])            # indexing is 0-based!
print("last  :", subjects[-1])           # negative index = from the end
print("slice :", rts[1:4])               # a slice: positions 1, 2 and 3, the end position 4 is left out
````

### 7.2 Methods and attributes

In Python, everything is an **object**: a number, a string, a list, a dictionary, everything (Python is an object oriented language). Think of an object as a thing in the real world: it can *do* something (its **methods**) and it can *have* properties (its **attributes**). Each object comes with two kinds of "tools" that you access through a `.` (the **dot operator**):

- **Methods** are *functions that belong to the object*. You call them *with* parentheses: `object.method_name(arguments)`.
- **Attributes** are *pieces of data stored inside the object*. You read them *without* parentheses: `object.attribute_name`.

This is the minimum "object-oriented programming" vocabulary you need to get through the rest of the course. You will see the `object.something()` pattern everywhere, again in section 13 when we meet `pandas` (`df.head()`, `df.describe()`), and from Week 2 onwards with `numpy` (`array.mean()`, `array.shape`).

Let's start with **list methods**. A list comes with a small toolbox for modifying itself *in place*:


````python
# --- List methods: modifying a list in place
subjects = ["S01", "S02", "S03"]        # the same three animals as above
print("start    :", subjects)

subjects.append("S04")                  # add an element at the end
print("append   :", subjects)

last = subjects.pop()                   # remove AND return the last element
print("pop      :", subjects, "→ removed:", last)

subjects.insert(1, "S00")               # insert at a given position
print("insert   :", subjects)

subjects.sort()                         # sort in place (alphabetically here)
print("sort     :", subjects)

subjects.reverse()                      # reverse in place
print("reverse  :", subjects)

# A method that *returns* a value rather than modifying in place:
print("count S01:", subjects.count("S01"))
````

Notice the difference between the two flavours of methods:

- `subjects.append("S04")` returns nothing: it *modifies the list in place*. Trying to write `result = subjects.append("S04")` would assign `None` to `result`, which is a *very* common beginner trap.
- `subjects.pop()` and `subjects.count(...)` *return a value* you can capture with `=`.

When in doubt, check the docs, or try it once with `print(object.method())` to see what comes back.

Now to **attributes**. Lists happen not to have any that you would use, but most other objects do, and the scientific libraries are full of them. A date is the clearest example: it *stores* a year and a month, and it can *work out* which weekday it falls on. The first two are attributes, the third is a method, and the parentheses are what tell them apart:


````python
# --- Attributes are read without parentheses, methods are called with them
from datetime import date

first_practical = date(2026, 9, 17)     # a date object: it stores a year, a month and a day

print("the date  :", first_practical)
print("its year  :", first_practical.year)      # attribute: data already inside the object
print("its month :", first_practical.month)     # attribute again, so no parentheses
print("weekday   :", first_practical.weekday()) # method: it works something out, so parentheses
print("(0 is Monday, so 3 means Thursday)")
````

**One trap worth knowing now.** Assigning a list to a second name does **not** copy it: both names point at the same list, so changing one changes the other. When you want an independent copy, ask for one with `.copy()`.

````python
# --- Two names for one list, and how to get a real copy
original = ["S01", "S02"]       # two animals again, the content does not matter here
same_list = original            # NOT a copy: another name for the same list
real_copy = original.copy()     # an independent copy

original.append("S03")

print("original :", original)
print("same_list:", same_list)   # changed too, because it is the same list
print("real_copy:", real_copy)   # unchanged
````

### 7.3 Dictionaries

## 8. Conditionals

A **conditional** runs different code depending on whether some condition is true. The basic structure is:
``` python
if condition_1:
    # runs only if condition_1 is True
elif condition_2:
    # runs only if condition_1 was False AND condition_2 is True
else:
    # runs if all the above were False
```

The `elif` and `else` branches are optional. Indentation matters in Python: the body of each branch must be **indented** (press `tab` on your keyboard to indent).

<img src="assets/01/ifelse.png" width="470">

*The conditions are checked in order. The first one that is true runs its own block and the rest are skipped, which is why the order you write them in matters. If none is true, the `else` block runs.*

### Comparison and logical operators

| Operator | Meaning                  | Example         |
|----------|--------------------------|-----------------|
| `==`     | equal to                 | `x == 0`        |
| `!=`     | not equal to             | `x != 0`        |
| `<`, `>` | less / greater than      | `x > 0.5`       |
| `<=`, `>=` | less / greater or equal | `x <= 1`       |
| `and`    | both must be true        | `0 < x and x < 1` |
| `or`     | at least one must be true | `x < 0 or x > 1` |
| `not`    | flips a boolean          | `not x > 0`     |


Before we write our first `if`, one word about what a comparison actually produces. `contrast > 0` is
not a question Python asks: it is an **expression that returns a value**, either `True` or `False`.
That value is a **boolean**, the fourth type from section 5, and you can store it, print it and

## 9. Loops

A **loop** repeats the same operation many times over:

- a `for` loop repeats an action for each item in a sequence;
- a `while` loop repeats an action **as long as a condition is true**.

<img src="assets/01/forwhileloop.jpeg" width="620">

*The difference in one picture. A `for` loop stops when it runs out of items, a `while` loop stops when its condition becomes false, which is why a `while` whose condition never changes runs forever.*

### 9.1 `for` loops

The most common loop iterates over a list (or any "iterable"):


````python
# --- Loop over a list and process each element
subjects = ["S01", "S02", "S03"]    # three animals
for subj in subjects:
    print(f"now processing subject {subj}")
````

Often you want to loop over a range of integers. `range(n)` produces the integers `0, 1, …, n-1`. `range(start, stop)` produces `start, …, stop-1`. `range(start, stop, step)` adds a stride.

````python
# --- range()
for i in range(5):
    print(i)
print("---")
for i in range(2, 10, 2):    # 2, 4, 6, 8
    print(i)
````

Use `enumerate()` when you need **both** the index and the value:

````python
# --- enumerate()
subjects = ["S01", "S02", "S03"]
for i, subj in enumerate(subjects):
    print(f"#{i}: {subj}")
````

Use `zip()` when you need to iterate over two lists in parallel:

````python
# --- zip()
subjects = ["S01", "S02", "S03"]    # three subjects
ages = [24, 31, 28]                 # and their ages, in years
for subj, age in zip(subjects, ages):
    print(f"{subj} is {age} years old.")
````

### 9.2 `while` loops

A `while` loop keeps repeating *as long as* a condition is true. Use it when you don't know in advance how many iterations you need:

````python
# --- while loop: keep halving until below a threshold
x = 1.0
n_iter = 0
while x > 0.01:
    x = x / 2
    n_iter += 1   # short for: n_iter = n_iter + 1
print(f"reached x = {x:.4f} in {n_iter} iterations")
````

### 9.3 Shorthand: list comprehensions 🔬

> 🔬 **Optional.** In real Python code, and later in the notebooks you will very quickly meet a more compact and practical way to build a list out of a `for` loop: a **list comprehension**. You will never have to write one from scratch in this course, but they do appear in code you read from Week 2 on, so it is worth learning to recognise the shape even if you skip the rest of this section.

The shape is:

``` python
[ expression   for   variable   in   iterable   if   condition ]
```

The `if condition` part is optional. Read it left to right: *"collect every value of `expression`, for every `variable` in `iterable`, keeping only those for which `condition` is true."*

There is nothing magical about it: a comprehension is just a shorter way to write a `for` loop that builds a list. Use it when it makes the code clearer; stick with a regular `for` loop when the body is more than one expression or when the logic is hard to read on a single line.


````python
# --- Three ways to do the same thing, plus a dict-comprehension bonus

# (1) Classic for-loop building a list
squares = []
for n in range(10):
    squares.append(n ** 2)
print("for-loop      :", squares)

# (2) Equivalent list comprehension
squares = [n ** 2 for n in range(10)]
print("comprehension :", squares)

# (3) Comprehension with a filter: only even squares
even_squares = [n ** 2 for n in range(10) if n % 2 == 0]
print("filtered      :", even_squares)

# (4) Dict comprehensions exist too: the same syntax, with a key: value pattern
square_table = {n: n ** 2 for n in range(5)}
print("dict comp     :", square_table)
````

## 10. Reading errors and asking for help

When your code goes wrong, Python prints a **traceback**: a multi-line error message ending with the *most informative* line. The golden rule is:

> **Read the traceback from the bottom up.** The last line tells you *what* went wrong; the lines above tell you *where*.

Let's deliberately cause five common errors and read each one carefully. (Each cell will *fail*, that's the point. Look at the error, then move on.)


````python
# --- NameError: using a name that was never defined
print(my_undefined_variable)
````

`NameError: name 'my_undefined_variable' is not defined`. Python looked for a variable called `my_undefined_variable` in the current kernel and couldn't find one. Either you mistyped the name, or you haven't yet run the cell that defines it. **Most common cause:** running cells out of order. *Restart Kernel → Run All* usually fixes it.

````python
# --- TypeError: combining incompatible types
"trial " + 3
````

`TypeError: can only concatenate str (not "int") to str`. Python's `+` operator means "addition" for numbers and "concatenation" for strings, but the two cannot be mixed. The fix here is to convert the integer to a string first: `"trial " + str(3)`. Even better, use an f-string: `f"trial {3}"`.

````python
# --- IndexError: asking for a position that doesn't exist
subjects = ["S01", "S02", "S03"]
print(subjects[5])
````

`IndexError: list index out of range`. The list has 3 elements (indices 0, 1, 2), so asking for index 5 makes no sense. **Most common cause:** off-by-one. Remember that Python indexing is *0-based*: the last element of a list of length `n` is at index `n - 1` (or, equivalently, `-1`).

````python
# --- SyntaxError: a line that Python cannot even read
# The colon at the end of an "if" line is not optional.
if 3 > 2
    print("three is bigger")
````

`SyntaxError: expected ':'`. This one is different from the three above. Python did not even start running the cell: it could not make sense of the text. **Nothing in the cell runs**, not even the lines before the mistake. Look at the line the message points to, and at the one just above it, since a missing bracket or quote often only becomes visible on the next line.

````python
# --- IndentationError: a block that is not indented
for subject in ["S01", "S02"]:
print(subject)
````

`IndentationError: expected an indented block`. In Python the indentation **is** the syntax. Everything that belongs inside an `if`, a `for` or a `while` has to be indented under it, by one tab or four spaces, consistently. This is the other error you will meet while typing your first loops.

### 13.3 Rows and columns with `.loc` and `.iloc`

Most of the time you want a *piece* of the table: some rows, some columns, or only the rows that meet a condition. pandas gives you two tools, and telling them apart is the single most common source of confusion for beginners.

`.iloc` selects by **position**: "the 3rd row and the 5th column", counting from 0, exactly like list indexing. `.loc` selects by **label**: "the row labelled 3 and the column called `lab`". Both take **rows first, columns second**.

| | `.iloc` | `.loc` |
|---|---------|--------|
| **Mnemonic** | **i** for *integer* | **l** for *label* |
| **Rows** | integer positions: `0`, `5`, `0:10`, `-1` | index labels: `0`, `5`, `0:10` |
| **Columns** | integer positions: `2`, `0:4` | column **names**: `"lab"`, `["lab", "sex"]` |
| **Slice `a:b`** | **excludes** `b`, so `iloc[0:3]` gives 3 rows | **includes** `b`, so `loc[0:3]` gives 4 rows |

They look interchangeable at first, because a freshly loaded table has the index `0, 1, 2, …`, so a row's position and its label are the same number. They stop being interchangeable as soon as you filter rows, because filtering keeps the original labels.

> **Rule of thumb.** Use `.loc` by default: naming a column says what you mean, and your code keeps working if someone adds a column upstream. Use `.iloc` when position is genuinely what you mean, such as "the first row" or "the last ten". Both of them also accept a condition in place of the row numbers, which is how you filter a table, and Week 3 does exactly that.

````python
# --- .iloc works with positions
print("first row, by position:")
print(mice.iloc[0])
````

````python
# --- .loc works with labels: row label 0, column called "lab"
print("one cell, by label   :", mice.loc[0, "lab"])
print("one cell, by position:", mice.iloc[0, 3])

# rows first, columns second
mice.loc[0:2, ["subject_id", "lab", "sex"]]
````

Time to use everything you have learned so far (variables, lists, dictionaries, conditionals, loops) on a very tiny dataset. Below are 12 trials from three of the mice you met in section 13, written straight into the notebook as a **list of dictionaries**. Each dictionary is one trial: which mouse, which stimulus, what it answered, whether that was correct, and how long it took.

Two of the trials have `no_go` set to `True`: the mouse did not answer in time, so there is no `correct` value for them. Missing values like these are normal in real data, and how you handle them is a decision you have to make in every analysis.

In section 13 you loaded a real table with pandas. Here the goal is the opposite: to *feel* what the same kind of analysis looks like when you write every step by hand.


````python
# --- A tiny dataset of 12 trials, hard-coded as a list of dictionaries
# One dictionary per trial. "contrast" is the stimulus, its sign giving the side and its size how
# visible it was; "response" is the side the mouse chose; "no_go" means it did not answer at all;
# "rt_s" is the reaction time in seconds.
trials = [
    {"mouse": "CSHL045", "contrast":  0.50, "response": "right", "correct": True,  "no_go": False, "rt_s": 0.42},
    {"mouse": "CSHL045", "contrast": -0.25, "response": "left",  "correct": True,  "no_go": False, "rt_s": 0.55},
    {"mouse": "CSHL045", "contrast":  0.125,"response": "right", "correct": True,  "no_go": False, "rt_s": 0.71},
    {"mouse": "CSHL045", "contrast": -0.0625,"response":"right", "correct": False, "no_go": False, "rt_s": 0.89},
    {"mouse": "CSHL046", "contrast":  0.25, "response": "right", "correct": True,  "no_go": False, "rt_s": 0.39},
    {"mouse": "CSHL046", "contrast": -0.50, "response": "left",  "correct": True,  "no_go": False, "rt_s": 0.44},
    {"mouse": "CSHL046", "contrast":  0.00, "response": None,    "correct": None,  "no_go": True,  "rt_s": 0.95},
    {"mouse": "CSHL046", "contrast": -0.125,"response": "right", "correct": False, "no_go": False, "rt_s": 0.78},
    {"mouse": "CSP001",  "contrast":  0.50, "response": "right", "correct": True,  "no_go": False, "rt_s": 0.38},
    {"mouse": "CSP001",  "contrast":  0.0625,"response":"left",  "correct": False, "no_go": False, "rt_s": 0.66},
    {"mouse": "CSP001",  "contrast": -0.25, "response": "left",  "correct": True,  "no_go": False, "rt_s": 0.52},
    {"mouse": "CSP001",  "contrast":  0.00, "response": None,    "correct": None,  "no_go": True,  "rt_s": 0.81},
]

print(f"Total trials       : {len(trials)}")
print(f"Keys of each trial : {list(trials[0].keys())}")
print(f"First trial        : {trials[0]}")
````

Let's count how many trials each mouse contributed, using a dictionary as a running tally. The pattern is: *start at zero, then for each trial, add one to the right counter*.


````python
# --- Count trials per mouse
trials_per_mouse = {}
for trial in trials:
    mouse = trial["mouse"]
    if mouse not in trials_per_mouse:
        trials_per_mouse[mouse] = 0
    trials_per_mouse[mouse] += 1

for mouse, n in trials_per_mouse.items():
    print(f"{mouse}: {n} trials")
````

And the mean reaction time of one mouse, computed by hand: sum first, then divide.


````python
# --- Mean reaction time of one mouse
total_rt = 0.0
n = 0
for trial in trials:
    if trial["mouse"] == "CSHL045":
        total_rt += trial["rt_s"]
        n += 1

mean_rt = total_rt / n
print(f"CSHL045: mean RT over {n} trials = {mean_rt:.3f} s")
````
