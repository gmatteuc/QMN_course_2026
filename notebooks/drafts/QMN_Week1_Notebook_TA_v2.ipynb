The data come from a **two-alternative forced choice (2AFC)** perceptual decision task.
On each trial a mouse sees a visual stimulus and reports which side it appeared on
by turning a wheel.

Stimulus strength is given by **coherence**, which is *signed*:
- the **sign** encodes the side the stimulus was on (negative = left, positive = right)
- the **magnitude** encodes how easy the trial was (0.5 = very easy, 0.05 = very hard)

A coherence of exactly `0.0` is a **catch trial**: no stimulus was shown on either side,
so there is no correct answer. This is why the `correct` column has missing values,
and it matters in Section 6.

### 2.2 ⭐️ Booleans

A **boolean** holds exactly one of two values: `True` or `False`.
Booleans are the result of comparisons
and are the backbone of all conditional logic.

| Boolean | Meaning | Example |
|---------|---------|---------|
| `True` | The logical truth | `5 > 3` |
| `False` | The logical falsity | `5 < 3` |

<br>

| Comparison sign | Meaning | Example |
|-----------------|---------|---------|
| `==` | Equal to | `5 == 5` |
| `!=` | Not equal to | `5 != 3` |

| `<=` | Less than or equal to | `5 <= 5` |

<br>

You can combine comparisons with **logical operators**:

**And**: `&` --> `True` if both sides are `True`, else `False` (i.e. 5 is greater than 3 **and** 5 is less than 10) <br>

**Not**: `~` --> `True` if the condition is `False`, else `False` (i.e. 5 is not greater than 3)

### 2.3 ⭐️ Lists

A **list** stores an ordered, **mutable** (that can be changed) sequence of values.
Lists are the most common container in Python for holding sequences of data.

Key operations covered below: creating, indexing, slicing, `append()`, `insert()`,
`remove()`, `pop()`, `sort()`, `reverse()`, and list comprehensions.


````python
# Creating a list, indexing, slicing, and len().

contrasts = [0.0, 0.06, 0.12, 0.25, 1.0]   # stimulus contrast levels
mouse_ids = ["mouse_01", "mouse_02", "mouse_03"]

# Indexing (Python starts at 0)
print("First contrast:", contrasts[0])
print("Last contrast: ", contrasts[-1])

# Slicing
print("Middle contrasts:", contrasts[1:4])

# Length
print("Number of contrast levels:", len(contrasts))
````

````python
# append() and insert() — adding items to a list.

# append() adds to the END of the list
contrasts.append(0.5)
print("After append(0.5):", contrasts)

# insert(index, value) adds at a SPECIFIC position
contrasts.insert(0, -1.0)   # insert -1.0 at position 0 (front)
print("After insert(0, -1.0):", contrasts)
````

### 2.3.1 ⭐️⭐️ List Comprehensions

A **list comprehension** builds a new list from an existing one in a single line.
It replaces the pattern of creating an empty list and appending inside a `for` loop.

```python
# Standard for-loop approach
result = []
for item in sequence:
    result.append(some_function(item))

# Equivalent list comprehension
result = [some_function(item) for item in sequence]
```

You can also add a condition to filter items:
```python
result = [item for item in sequence if condition]
```


````python
# List comprehensions — concise list building.

contrasts = [0.0, 0.06, 0.12, 0.25, 1.0]

# Convert contrasts from proportions to percentages
contrasts_pct = [c * 100 for c in contrasts]
print("Contrasts as %:", contrasts_pct)

# Keep only visible contrasts (exclude catch trials where contrast == 0)
visible = [c for c in contrasts if c > 0.0]
print("Visible contrasts:", visible)

# Classify each contrast in one line
labels = ["catch" if c == 0.0 else "visible" for c in contrasts]
print("Labels:", labels)
````

### 4.5 ⭐️⭐️ Accessing Rows and Columns with `.loc` and `.iloc`

So far we have selected whole columns with `df["rt_s"]`. But most of the time you want a
*piece* of the table: certain rows, certain columns, or only the rows that meet a condition.
pandas gives you two tools for this, and the difference between them is the single most
common source of confusion for beginners.

`.iloc` selects by **position** meaning : "give me the 3rd row and the 5th column", counting from 0,
exactly like list indexing. `.loc` selects by **label** meaning : "give me the row labelled 3 and the
column called `rt_s`". Both take **rows first, columns second**: `df.iloc[rows, columns]`
and `df.loc[rows, columns]`.

They look interchangeable at first because a freshly loaded table has an index of
`0, 1, 2, 3, …`, so a row's position and its label are the same number. They stop being
interchangeable as soon as you filter rows, because filtering keeps the original labels.

| | `.iloc` | `.loc` |
|---|---------|--------|
| **Mnemonic** | **i** for *integer* | **l** for *label* |
| **Expects for rows** | integer positions: `0`, `5`, `0:10`, `-1` | index labels: `0`, `5`, `0:10` |
| **Expects for columns** | integer positions: `2`, `0:4` | column **names**: `"rt_s"`, `["rt_s", "session"]` |
| **Slice `a:b`** | **excludes** `b` — `iloc[0:3]` gives 3 rows | **includes** `b` — `loc[0:3]` gives 4 rows |
| **Returns** | Series if you ask for one column, DataFrame otherwise | same |
| **You must specify** | *where the data sits* (its number) | *what the data is called* (its name) |

<br>

> 📌 **Rule of thumb.** Use `.loc` by default: naming a column is self-documenting, and your
> code keeps working if someone inserts a column upstream. Use `.iloc` only when position is
> genuinely what you mean (e.g."the first row", "the last 10 rows").

````python
# .iloc — selection by POSITION. Everything is a number, counting from 0.

# 1. A single column, by its position. Column 7 is the last one.
print("Column at position 7 is:", df.columns[7])
df.iloc[:, 7].head(3)       # ":" means "all rows"
````

````python
# 3. Rows AND columns together: rows first, columns second.
print("--- first 5 rows, first 4 columns ---")
df.iloc[0:5, 0:4]
````

### 5.1 ⭐️⭐️ For Loops and Conditionals

A `for` loop iterates over each item in a sequence. Combined with `if / elif / else`, it lets you classify, filter, and aggregate data row by row.

````python
# Classify each contrast level and store the label in a dictionary.

contrasts = [0.0, 0.06, 0.12, 0.25, 1.0]
contrast_labels = {}

for c in contrasts:
    if c == 0.0:
        label = "catch trial"
    elif c < 0.12:
        label = "low contrast"
    elif c < 0.5:
        label = "medium contrast"
    else:
        label = "high contrast"
    contrast_labels[c] = label

for contrast, label in contrast_labels.items():
    print(f"Contrast {contrast:.2f} → {label}")
````

### 5.1.1 ⭐️ Looping Over Positions with `range()`

Sometimes you need an item's **position**, not just the item itself — for instance when reading two lists side by side, or when you want to count iterations.

`range(n)` produces the whole numbers from `0` up to **but not including** `n`. It pairs naturally with `len()`:

```python
for i in range(len(my_list)):
    ...
```

This gives you `i` as the index, so `my_list[i]` is the item and `other_list[i]` is the matching item in a second list.

### 5.3 ⭐️⭐️ While Loops

A `while` loop keeps running until its condition becomes `False`.
Use it when you don't know in advance how many iterations you need —
for example, when sampling until a threshold is met, or processing a queue of items.

> ⚠️ Always make sure the condition will eventually become `False`, otherwise the loop runs forever.

````python
# While loop — how many trials before this mouse responds within 1 second?
# This mimics an adaptive stopping rule used in real behavioural experiments.

import random
random.seed(42)   # fix the random seed so results are reproducible

response_times = [2.1, 1.8, 1.5, 0.9, 1.2, 0.7]   # response times in seconds with one per trial

i = 0
while response_times[i] >= 1.0:
    i = i + 1

print("First fast enough response on trial", i, "(", response_times[i], "s)")
````

## 6. ⭐️⭐️⭐️ Capstone Project — Putting It All Together

````python
# TODO: Build a two-stage pipeline that cleans the dataset, saves it, reloads it,
# and reports which sessions fail a quality check.
# Do not use .groupby(), .value_counts() or .mean() try and build the counts with loops.

# STAGE 1 — filter and save a checkpoint
# 1. Build a boolean mask keeping trials that are NOT catch trials (coherence != 0.0)
#    AND whose 'correct' entry is not missing. Wrap each condition in its own parentheses.
# 2. Apply the mask while keeping only these columns:
#    subject_id, session, coherence, correct, rt_s
# 3. Reset the index with the .reset_index() method with argument drop=True, so that position and label agree again.
# 4. Save the result to the "trials_clean.csv" file without the index column using the .to_csv() method and as argument index=False.
# 5. Print how many rows you started with, how many you removed, and how many remain.

# STAGE 2 — reload the checkpoint
# 6. Read "trials_clean.csv" back into a NEW variable. Do not reuse the table from Stage 1 !
# 7. Print its shape and its column names to confirm the file survived the round trip.

# STAGE 3 — score each session
# 8. A session is identified by the PAIR (subject_id, session), because session numbers
#    restart at 1 for every mouse. A tuple makes a good dictionary key.
#    Loop over the reloaded trials and count, for each session:
#    the total number of trials, and how many pass the check (rt_s < 1.0).
# 9. Compute a pass rate for each session.
# 10. Using an if/else inside a loop, split the sessions into two lists:
#     "passed" if the pass rate is at least 85%, "flagged" otherwise.
# 11. Print how many sessions passed and how many were flagged.

# STAGE 4 — export only the flagged sessions
# 12. Build a DataFrame containing ONLY the flagged sessions, with the columns:
#     mouse_id, session, n_trials, pass_rate
#     Sort it from worst to best.
# 13. Save it to "flagged_sessions.csv" without the index, then read it back and print it.
# 14. In a comment, state any pattern you notice in which sessions were flagged.
````

````python
# ---------------------------------------------------------------
# STAGE 1 — filter rows and columns, then save the result to a csv file
# ---------------------------------------------------------------
# Each condition needs its own parentheses; & means "and".
mask_usable = (df['coherence'] != 0.0) & (df['correct'].notna())

my_columns = ['subject_id', 'session', 'coherence', 'correct', 'rt_s']

df_clean = df.loc[mask_usable, my_columns]      # rows from the mask, columns from the list
df_clean = df_clean.reset_index(drop=True)      # renumber 0, 1, 2, ... with no gaps

df_clean.to_csv("trials_clean.csv", index=False) # save without the index column, will save to the location of the notebook

print("STAGE 1 — cleaning")
print("  rows at the start:", len(df))
print("  rows removed:     ", len(df) - len(df_clean))
print("  rows remaining:   ", len(df_clean))
print("  saved to trials_clean.csv")

# ---------------------------------------------------------------
# STAGE 2 — reload the checkpoint into a NEW variable
# ---------------------------------------------------------------
df_reloaded = pd.read_csv("trials_clean.csv")

print("\nSTAGE 2 — reload")
print("  shape:  ", df_reloaded.shape)
print("  columns:", df_reloaded.columns.tolist())

# ---------------------------------------------------------------
# STAGE 3 — score each session
# ---------------------------------------------------------------
subject_col = df_reloaded['subject_id'].tolist()
session_col = df_reloaded['session'].tolist()
rt_col      = df_reloaded['rt_s'].tolist()

n_trials_in = {}
n_passed_in = {}

for i in range(len(subject_col)):
    key = (subject_col[i], session_col[i])      # a tuple, e.g. ("S02", 3)

    if key in n_trials_in:
        n_trials_in[key] = n_trials_in[key] + 1
    else:
        n_trials_in[key] = 1
        n_passed_in[key] = 0

    if rt_col[i] < 1.0:
        n_passed_in[key] = n_passed_in[key] + 1

# Turn the counts into a pass rate
pass_rate_in = {}
for key in n_trials_in:
    pass_rate_in[key] = n_passed_in[key] / n_trials_in[key]

# Split into two lists with an if/else
passed  = []
flagged = []
for key in pass_rate_in:
    if pass_rate_in[key] >= 0.85:
        passed.append(key)
    else:
        flagged.append(key)

print("\nSTAGE 3 — scoring")
print("  sessions found:", len(pass_rate_in))
print("  passed: ", len(passed))
print("  flagged:", len(flagged))

# ---------------------------------------------------------------
# STAGE 4 — export only the flagged sessions
# ---------------------------------------------------------------

sortable = []
for key in flagged:
    sortable.append([pass_rate_in[key], key])           # pair the pass rate with the session key so we can sort by rate

sortable.sort()

flagged_sorted = []
for pair in sortable:
    flagged_sorted.append(pair[1])                      # pair[0] is the rate, pair[1] is the session

print("\nSorted flagged sessions from worst to best:", flagged_sorted)

rows = []
for key in flagged_sorted:
    rows.append({
        "mouse_id":  key[0],                            # the tuple unpacks into its two parts
        "session":   key[1],
        "n_trials":  n_trials_in[key],
        "pass_rate": round(pass_rate_in[key], 4),
    })

print("\nFlagged sessions (worst to best):")
for row in rows:
   print(row)

flagged_table = pd.DataFrame(rows)
flagged_table.to_csv("flagged_sessions.csv", index=False)

flagged_reloaded = pd.read_csv("flagged_sessions.csv")

print("\nSTAGE 4 — flagged sessions")
print(flagged_reloaded)
print("\nround trip intact:", flagged_table.equals(flagged_reloaded))

# 14. Five of the six flagged sessions belong to S02, so this is mostly one slow
#     animal rather than a scattered problem. The other flagged session is S01's
#     session 1, and S02's two worst sessions are also its first two — early
#     sessions look worse than later ones, which is what you would expect if the
#     animals were still learning the task.
````
